### Fix: Invalid API Key

The error "API key not valid" indicates that the `API_KEY` used in the code is not a valid Gemini API key. You need to obtain a valid key and store it securely.

**Steps to fix:**
1.  **Get a Gemini API Key:** Go to [Google AI Studio](https://aistudio.google.com/app/apikey) and create an API key.
2.  **Store the API Key Securely in Colab:**
    *   In the left sidebar of Colab, click the "🔑" (Secrets) icon.
    *   Click "+ New secret".
    *   For the "Name" field, enter `GOOGLE_API_KEY`.
    *   For the "Value" field, paste your actual Gemini API key.
    *   Make sure "Notebook access" is toggled on for this notebook.

After setting up your API key in Colab Secrets, the following code will retrieve and use it.

```python
from google.colab import userdata
import google.generativeai as genai

# Retrieve the API key from Colab Secrets
API_KEY = userdata.get('GOOGLE_API_KEY')

# Configure the genai library with your API key
genai.configure(api_key=API_KEY)

# Re-create the chat session with the valid API key
client = genai.Client(api_key=API_KEY)
chat = client.chats.create(
    model="gemini-2.5-flash"
)

print("Gemini Chatbot")
print("Type 'exit' to stop.\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Chat ended.")
        break

    response = chat.send_message(user_input)

    print("Bot:", response.text)
    print()
```
